```python
import pickle
import pandas as pd
from IPython.display import display
with open("./student_performance_decision_tree_pipeline.bin","rb") as File:
    student_grade_predictor = pickle.load(File)
with open("./student_performance_preprocessor.bin","rb") as File1:
    preprocessor = pickle.load(File1)
grade_class_mapping = {0.0: 'A', 1.0: 'B', 2.0: 'C', 3.0: 'D', 4.0: 'F'}
```

```python
student_input = pd.DataFrame([{
    'Age': 20,
    'Gender': 0,
    'Ethnicity': 0,
    'ParentalEducation': 4,
    'StudyTimeWeekly': 6.0,
    'Absences': 2,
    'Tutoring': 1,
    'ParentalSupport': 3,
    'Extracurricular': 0,
    'Sports': 0,
    'Music': 0,
    'Volunteering': 0
}])
student_input
```

```python
processed_input = preprocessor.transform(student_input)
processed_input
```

```python
print("Original features:", student_input.shape[1])
print("Processed features:", processed_input.shape[1])
```

```python
predicted_output = student_grade_predictor.predict(processed_input)
grade_class_mapping[predicted_output[0]]
```
